```python
import pandas as pd

def get_input_descriptor():
    filename = input("Which file do you want to open?: ")
    df = pd.read_csv(filename)
    return df

def get_data_list(df, column_name):
    column_name = column_name.strip().title()
    data_list = []
    for index, row in df.iterrows():
        date_val = str(row['Date'])
        col_val = row[column_name]
        data_list.append([date_val, col_val])   
    return data_list

def get_month_stats(list_of_lists, date_string):
    parts = date_string.split('/')
    target_month = parts[0]
    target_year = parts[1]
    if len(target_month) == 1:
        target_month = '0' + target_month
        
    matching_rows = []
    for row in list_of_lists:
        row_date = row[0]
        row_date_parts = row_date.split('-')
        row_year = row_date_parts[0]
        row_month = row_date_parts[1]
        if row_year == target_year and row_month == target_month:
            matching_rows.append(row)   
    total_sum = 0
    count = 0
    for row in matching_rows:
        total_sum = total_sum + row[1]
        count = count + 1
    if count > 0:
        average_val = total_sum / count
    else:
        average_val = 0
    print("The average for", date_string, "is:", average_val)
    matching_rows.sort(key=lambda x: x[1])
    print("Minimum 3 days are:")
    for i in range(3):
        if i < len(matching_rows):
            print("Date:", matching_rows[i][0], "Value:", matching_rows[i][1])     
    print("Maximum 3 days are:")
    for i in range(1, 4):
        if len(matching_rows) - i >= 0:
            idx = len(matching_rows) - i
            print("Date:", matching_rows[idx][0], "Value:", matching_rows[idx][1])


def get_all_stats(list_of_lists):
    monthly_data = {}
    for row in list_of_lists:
        row_date = row[0] 
        row_parts = row_date.split('-')
        year_month = row_parts[0] + "-" + row_parts[1]
        val = row[1]
        if year_month not in monthly_data:
            monthly_data[year_month] = [0.0, 0]
        monthly_data[year_month][0] = monthly_data[year_month][0] + val
        monthly_data[year_month][1] = monthly_data[year_month][1] + 1
        
    averages_list = []
    for ym in monthly_data:
        total = monthly_data[ym][0]
        count = monthly_data[ym][1]
        avg = total / count
        averages_list.append([ym, avg])
        
    averages_list.sort(key=lambda x: x[1])
    print('\nBOTTOM 3 monthly averages are:')
    for i in range(3):
        if i < len(averages_list):
            print("Date:", averages_list[i][0], "Value:", averages_list[i][1])
            
    print("\nTOP 3 monthly averages are:")
    for i in range(1, 4):
        if len(averages_list) - i >= 0:
            idx = len(averages_list) - i
            print("Date:", averages_list[idx][0], "Value:", averages_list[idx][1])

def main():
    running = True
    while running:
        df = get_input_descriptor()
        col_name = input("Which column to get information about? (Type Open, High, Low, Close, Adj Close, or Volume): ")
        data_list = get_data_list(df, col_name)
        choice = input("Stats for a specific month or all months? (Type 1 for a specific month, 2 for all): ")
        
        if choice == '1':
            month_input = input("Which month/year?: ")
            get_month_stats(data_list, month_input)
        elif choice == '2':
            get_all_stats(data_list)
        else:
            print("Invalid choice.")
            
        again = input("Run again? (Y/N): ")
        if again.strip().upper() != 'Y':
            running = False
            print("Thanks for using this stock analysis program.")

if __name__ == "__main__":
    main()
```
